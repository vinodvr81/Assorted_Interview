check valid sequence in order and return either true or false ex: {([])} -> valid sequence {[(])} -> invalid sequence

```python
def check_valid_sequence(given_string: str) -> bool:
    check_tup = [("{","}"),("[","]"),("(",")")]
    if len(given_string)%2 == 0:
        for count in range(0, round(len(given_string)/2)):
            first_half = list(given_string)[:round(len(given_string)/2)]
            second_half = list(reversed(list(given_string)[round(len(given_string)/2):]))
            if (first_half[count],second_half[count]) in check_tup:
                continue
            else:
                return False
    else:
        return False
    return True
```
