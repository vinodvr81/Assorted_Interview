check valid sequence in order and return either true or false ex: {([])} -> valid sequence {[(])} -> invalid sequence

```python
def check_valid_sequence(given_string: str) -> bool:
    check_tup = [("{","}"),("[","]"),("(",")")]
    if len(given_string)%2 == 0:
        for count in range(0, round(len(given_string)/2)):
            first_half = list(given_string)[:round(len(given_string)/2)]
            second_half = list(reversed(list(given_string)[round(len(given_string)/2):]))
            if (first_half[count],second_half[count]) in check_tup:
                continue
            else:
                return False
    else:
        return False
    return True
```

```python
print(check_valid_sequence("{([])}"))
print(check_valid_sequence("{[(])}"))
print(check_valid_sequence("{[()]"))
```

Higest Alphabet in a string

```python
def highest_occurence_of_alphabet(given_string: str) -> tuple:
    string_dictionary: dict = {}
    for val in list(given_string):
        if val not in string_dictionary:
            string_dictionary[val] = 1
        else:
            string_dictionary[val] += 1
    higest_occurence = max(string_dictionary, key=string_dictionary.get)
    return higest_occurence, string_dictionary

    
```

```python
print(highest_occurence_of_alphabet("monsoon"))
```

flatten a list

```python
def check_string_to_integer(given_string: str):
    try:
        return int(given_string)
    except ValueError:
        print("Cannot convert this string to integer")

def flatten_list(given_list: list) -> list:
    flattened_list: list = []
    for val in given_list:
        if isinstance(val, (int, float)):
            flattened_list.append(val)
        elif isinstance(val, str):
            flattened_list.append(check_string_to_integer(val))
        elif isinstance(val, list):
            flattened_list.extend(flatten_list(val))
    return flattened_list

print(flatten_list([1,2,3,[4,5,"6","seven",8],[9,[10,11],12],13]))
            
```
